# API Textrazor : NLP et analyse de texte

## Imports

```python
import json
import requests
```

## Utiliser Textrazor pour extraire les entités nommées du corpus Camille

```python
# Récupération du fichier texte de CAMille
file = "../../data/all.txt"

with open(file, "r", encoding="utf-8") as reader:
    content = reader.read()

# Récupération de la clé API
with open("../../data/api.env", "r") as reader:
    api_key = reader.read().strip().split("=", 1)[1]
```

## Interroger l'API Textrazor

```python

url = "https://api.textrazor.com/"

headers = {'X-TextRazor-Key': api_key}

data = {
    'text': content[:5000],
    'extractors': 'entities,topics'
}

resp = requests.post(url, data=data, headers=headers)

print(resp.status_code)

result = json.loads(resp.text)

for item in result['response']['entities']:
    name = item['entityId']
    entity_type = item.get('type', [])

    print(f"{name} : {entity_type}")
```
